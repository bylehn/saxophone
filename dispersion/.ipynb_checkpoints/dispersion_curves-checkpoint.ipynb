```python
import numpy as np
from matplotlib  import pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.colors as mcolors
import matplotlib.ticker as ticker
from scipy.spatial import Delaunay
import scipy.linalg as scla
import networkx as nx
import time

from dispersion import GetBranchedBro

def createDelaunayGraph_OG(NS,rseed,r_c,del_x):
    np.random.seed(rseed)
    xm, ym = np.meshgrid(np.arange(1, NS+1), np.arange(1, NS+1))
    X = np.vstack((xm.flatten(), ym.flatten())).T
    N = X.shape[0]
    X = X + del_x * 2 * (0.5 - np.random.rand(N, 2))
    
    DT = Delaunay(X)
    ET = np.empty((0, 2), dtype=int)
    for T in DT.simplices:
        ET = np.vstack((ET, [T[0], T[1]], [T[1], T[2]], [T[0], T[2]]))
    
    ET=np.sort(ET)
    R=np.linalg.norm(X[ET[:, 0], :] - X[ET[:, 1], :],axis=1)
    EN=ET[R<r_c,:]
    A = np.zeros((N, N))
    A[EN[:,0],EN[:,1]]=1
    
    
    G = nx.Graph(A)
    E=np.array(G.edges)
    
    return N,G,X,E

def createDelaunayGraph(NS, rseed, r_c, del_x, periodic=False):
    np.random.seed(rseed)
    xm, ym = np.meshgrid(np.arange(1, NS+1), np.arange(1, NS+1))
    X = np.vstack((xm.flatten(), ym.flatten())).T
    N = X.shape[0]
    X = X + del_x * 2 * (0.5 - np.random.rand(N, 2))
    if not periodic:
        X = np.clip(X, 1, NS)
        DT = Delaunay(X)
        ET = []
        for T in DT.simplices:
            ET.append([T[0], T[1]])
            ET.append([T[1], T[2]])
            ET.append([T[0], T[2]])
        ET = np.sort(np.array(ET), axis=1)
        R = np.linalg.norm(X[ET[:,0]] - X[ET[:,1]], axis=1)
        EN = ET[R < r_c]
        A = np.zeros((N, N), dtype=int)
        for e0, e1 in EN:
            A[e0, e1] = 1
        G = nx.Graph(A)
        E = np.array(G.edges())
        return N, G, X, E, NS
    else:
        L = NS
        X -= 1
        shifts = []
        for i in [-1, 0, 1]:
            for j in [-1, 0, 1]:
                shifts.append([i * L, j * L])
        shifts = np.array(shifts)
        tiled_points = []
        indices = []
        for s in shifts:
            tiled_points.append(X + s)
            indices.append(np.arange(N))
        tiled_points = np.vstack(tiled_points)
        indices = np.hstack(indices)
        DT = Delaunay(tiled_points)
        def tri_circumcenter(a, b, c):
            d = 2*(a[0]*(b[1]-c[1]) + b[0]*(c[1]-a[1]) + c[0]*(a[1]-b[1]))
            ux = ((np.sum(a*a)*(b[1]-c[1]) + np.sum(b*b)*(c[1]-a[1]) + np.sum(c*c)*(a[1]-b[1]))/d)
            uy = ((np.sum(a*a)*(c[0]-b[0]) + np.sum(b*b)*(a[0]-c[0]) + np.sum(c*c)*(b[0]-a[0]))/d)
            return np.array([ux, uy])
        edge_set = set()
        for simplex in DT.simplices:
            a = tiled_points[simplex[0]]
            b = tiled_points[simplex[1]]
            c = tiled_points[simplex[2]]
            cc = tri_circumcenter(a, b, c)
            if 0 <= cc[0] < L and 0 <= cc[1] < L:
                inds = [indices[simplex[0]], indices[simplex[1]], indices[simplex[2]]]
                for i in range(3):
                    i1 = inds[i]
                    i2 = inds[(i+1) % 3]
                    if i1 != i2:
                        edge_set.add(tuple(sorted((i1, i2))))
        G = nx.Graph()
        G.add_nodes_from(range(N))
        final_edges = []
        for (i0, i1) in edge_set:
            p0 = X[i0]
            p1 = X[i1]
            diff = p1 - p0
            diff -= L*np.rint(diff/L)
            dist = np.sqrt(np.sum(diff*diff))
            if dist < r_c:
                final_edges.append((i0, i1))
        G.add_edges_from(final_edges)
        E = np.array(list(G.edges()))
        X += 1
        return N, G, X, E, L

def createSquareGraph(NS,rseed,r_c,del_x, periodic=False):
    np.random.seed(rseed)
    xm, ym = np.meshgrid(np.arange(1, NS+1), np.arange(1, NS+1))
    X = np.vstack((xm.flatten(), ym.flatten())).T
    N = X.shape[0]
    
    
    T=np.tile(X,(N,1,1))
    T2=np.moveaxis(T,0,1)
    TD=T-T2
    if periodic:
        box_length = NS
        TD = (TD/box_length)
        TDN=TD-np.rint(TD)
        TD=box_length*TDN
    distances_all = np.linalg.norm(TD,axis=2)
    E1=np.array(np.where(distances_all<r_c))
    E=np.transpose(E1[:,E1[0]>E1[1]])
    
    G = nx.Graph()
    G.add_nodes_from(range(N))
    G.add_edges_from(E)
    E=np.array(G.edges)
    X = X+ del_x * 2 * (0.5 - np.random.rand(N, 2))
    if periodic:
        return N,G,X,E,box_length
    else:
        return N,G,X,E, NS

def createLineGraph(NS,rseed,r_c,del_x, periodic=False):
    np.random.seed(rseed)
    xm, ym = np.meshgrid(np.arange(1, NS+1), np.arange(1, 2))

    #x1, x2 = np.arange(1, NS + 1), np.arange(1.5, NS + 1.5)      
    #y1, y2 = np.ones_like(x1), np.ones_like(x2) * 2
    #xm, ym = np.concatenate((x1, x2)), np.concatenate((y1, y2))
    X = np.vstack((xm.flatten(), ym.flatten())).T
    N = X.shape[0]
    
    
    T=np.tile(X,(N,1,1))
    T2=np.moveaxis(T,0,1)
    TD=T-T2
    if periodic:
        box_length = NS
        TD = (TD/box_length)
        TDN=TD-np.rint(TD)
        TD=box_length*TDN
    distances_all = np.linalg.norm(TD,axis=2)
    E1=np.array(np.where(distances_all<r_c))
    E=np.transpose(E1[:,E1[0]>E1[1]])
    G = nx.Graph()
    G.add_nodes_from(range(N))
    G.add_edges_from(E)
    
    if NS<3:
        G.add_edge(0,NS-1)
        
    E=np.array(G.edges)
    X = X+ del_x * 2 * (0.5 - np.random.rand(N, 2))
    if periodic:
        return N,G,X,E,box_length
    else:
        return N,G,X,E

def getBondImportance(X,C,V,D,D_range):
    modes = np.where((D > D_range[0]) & (D < D_range[1]))[0]
    delta_E=C.T@V
    EC=delta_E[:,modes]
    bondImportance=np.mean(np.abs(EC),axis=1)
    return bondImportance/np.max(bondImportance)

def createCompatibility(N,X,E, box_length = None, periodic=False):
    
    N_b = E.shape[0]
    C = np.zeros((2 * N, N_b))
    
    for i in range(N_b):
        b_vec = X[E[i, 0], :] - X[E[i, 1], :]
        if periodic:
            b_vec_scaled = b_vec/box_length
            b_vec = box_length*(b_vec_scaled - np.rint(b_vec_scaled))
        b_vec = b_vec / np.linalg.norm(b_vec)
        for ki in range(2):
            j = E[i, ki]
            xind = 2 * j 
            C[xind:xind + 2, i] = ((-1) ** ki) * b_vec

    return C

def createSpringConstants(X,E,k_1, box_length=None, periodic = False):
    #k_1 is the spring constant for a spring of unit length
    displacements=X[E[:,0],:]-X[E[:,1],:]
    if periodic:
        displacements_scaled = displacements/box_length
        displacements = box_length*(displacements_scaled 
                                    - np.rint(displacements_scaled))
    distance=np.linalg.norm(displacements,axis=1)
    return k_1/distance,distance

def getForbiddenModes(C, k, M, w_c, dw):
    kd = np.diag(np.squeeze(k))
    K = C @ kd @ C.T
    DMAT = np.linalg.inv(M) @ K
    D, V = scla.eig(DMAT,overwrite_a=True)
    D = np.real(D)
    w=np.sqrt(np.abs(D))
    forbidden_states=np.sum(np.logical_and(w>w_c-dw/2,w<w_c+dw/2))
    V=np.real(V)
    return D, V,forbidden_states

def ageSprings(k_old,X,C,V,D,D_range,ageing_rate):
    bond_importance=getBondImportance(X,C,V,D,D_range)
    bond_importance_centered=bond_importance-np.mean(bond_importance)
    bond_importance_normalized=bond_importance_centered/np.max(np.abs(bond_importance_centered))

    k_new=k_old*(1+2*ageing_rate*bond_importance_normalized)
    return k_new
    
def optimizeAgeing(C, k, M, w_c, dw, N_trials,ageing_rate,success_frac):
    w_range=[w_c-dw/2,w_c+dw/2]
    D_range = [x**2 for x in w_range]
    D, V, forbidden_states_initial = getForbiddenModes(C, k, M, w_c, dw)
    if forbidden_states_initial==0:
        return k,1,0
    for trial in range(1, N_trials+1):
        k=ageSprings(k,X,C,V,D,D_range,ageing_rate)
    
        D, V, forbidden_states = getForbiddenModes(C, k, M, w_c, dw)
        if trial%100==0:
            print(trial,forbidden_states)
    
        if forbidden_states<=success_frac*forbidden_states_initial:
            
            return k, 1,trial
    
    return k,0,trial

def optimizeAgeing_q(C, q_points, k, M, w_c, dw, N_trials,ageing_rate,success_frac):
    w_range=[w_c-dw/2,w_c+dw/2]
    D_range = [x**2 for x in w_range]
    D, V, forbidden_states_initial, C_FT = getForbiddenModes_q(C, q_points, k, M, w_c, dw)
    #print(forbidden_states_initial)
    if forbidden_states_initial==0:
        return k,1,0
    for trial in range(1, N_trials+1):
        k=ageSprings_q(k,X,C,V,D,D_range,ageing_rate)
        D, V, forbidden_states, C_FT = getForbiddenModes_q(C, q_points, k, M, w_c, dw)
        if trial%100==0:
            print(f"{trial}\t{forbidden_states}\r")
        if forbidden_states<=success_frac*forbidden_states_initial:
            return k, 1,trial
    return k,0,trial

def getForbiddenModes_q(C, q_points, k, M, w_c, dw):
    forbidden_states = 0
    D=[]
    V=[]
    Cft=[]
    for i, q in enumerate(q_points):
        Cq = disp.FT_compatability(C, N, X, E, q)
        Di, Vi = disp.getDV(Cq, k, M)
        Di = np.abs(np.real(Di))
        Vi = np.real(Vi)
        D.append(Di)
        w=np.sqrt(np.abs(Di))
        for band in np.arange(0,2*N):
            forbidden_states+=np.sum(np.logical_and(w[band]>w_c-dw/2,w[band]<w_c+dw/2))
        V.append(Vi)
        Cft.append(Cq)
    return np.array(D), np.array(V),forbidden_states, np.array(Cft)

def ageSprings_q(k_old, X, C, V, D, D_range, ageing_rate):
    bond_importance_normalized = 0
    norm = 0
    for i, (Di, Vi) in enumerate(zip(D, V)):
        bond_importance = getBondImportance_q(X, C, Vi, Di, D_range)
        if bond_importance.size == 0:
            continue 
        bond_importance_centered = bond_importance - np.mean(bond_importance)
        max_abs_val = np.max(np.abs(bond_importance_centered))
        if max_abs_val != 0: 
            bond_importance_normalized += bond_importance_centered / max_abs_val
            norm += 1
    if norm > 0:
        bond_importance_normalized /= norm
    else:
        bond_importance_normalized = 0 
    k_new = k_old * (1 + 2 * ageing_rate * bond_importance_normalized)
    return k_new

def getBondImportance_q(X,C,V,D,D_range):
    modes = np.where((D > D_range[0]) & (D < D_range[1]))[0]
    if modes.size == 0:
        return np.array([]) #np.ones(np.shape(C.T)[0])
    #print(modes, D, D_range[0], D_range[1])
    delta_E=C.T@V
    #print(np.shape(C.T),np.shape(V),np.shape(D))
    EC=delta_E[:,modes]
    bondImportance=np.mean(np.abs(EC),axis=1)
    return bondImportance/np.max(bondImportance)

def set_ticks(axis, min_val, max_val):
    ticks = np.linspace(min_val, max_val, 4)
    axis.set_xticks(ticks)
    axis.set_xticklabels([f'{tick:.2f}' for tick in ticks])

def calculate_dos(disp, C, k2, M, N, X, E, num_q_points, num_bins):
    Lx = np.abs(X.T[0].max() - X.T[0].min())
    Ly = np.abs(X.T[1].max() - X.T[1].min())
    
    qmax_x = 2 * np.pi / Lx
    qmax_y = 2 * np.pi / Ly

    # Compute q_values for three paths
    q_values_x = disp.create_q_path(0, qmax_x, 0, 0, num_q_points)
    wx = disp.get_dispersion(C, k2, M, N, X, E, q_values_x)
    q_values_xy = disp.create_q_path(0, qmax_x, 0, qmax_y, num_q_points)
    wxy = disp.get_dispersion(C, k2, M, N, X, E, q_values_xy)
    q_values_y = disp.create_q_path(qmax_x, qmax_x, qmax_y, 0, num_q_points)
    wy = disp.get_dispersion(C, k2, M, N, X, E, q_values_y)

    # Combine the frequency data from all paths
    frequencies = np.hstack([wx.flatten(), wxy.flatten(), wy.flatten()])

    # Calculate DOS using histogram
    dos, bin_edges = np.histogram(frequencies, bins=num_bins, density=True)
    bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

    return bin_edges[1:], dos

def calculate_dos_rand(disp, C, k2, M, N, X, E, num_q_points, num_bins):
    Lx = np.abs(X.T[0].max() - X.T[0].min())
    Ly = np.abs(X.T[1].max() - X.T[1].min())
    
    qmax_x = 2 * np.pi / Lx
    qmax_y = 2 * np.pi / Ly

    # Compute q_values for three paths
    q_values = np.column_stack((np.random.uniform(0, qmax_x, num_q_points), np.random.uniform(0, qmax_x, num_q_points)))
    wx = disp.get_dispersion(C, k2, M, N, X, E, q_values)

    # Combine the frequency data from all paths
    frequencies = np.hstack([w.flatten()])

    # Calculate DOS using histogram
    dos, bin_edges = np.histogram(frequencies, bins=num_bins, density=True)
    bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

    return bin_edges[1:], dos, q_values

def calculate_dos_qx(disp, C, k2, M, N, X, E, num_q_points, num_bins):
    Lx = np.abs(X.T[0].max() - X.T[0].min())
    
    qmax_x = 2 * np.pi / Lx

    # Compute q_values for three paths
    q_values_x = disp.create_q_path(0, qmax_x, 0, 0, num_q_points)
    wx = disp.get_dispersion(C, k2, M, N, X, E, q_values_x)
    # Combine the frequency data from all paths
    frequencies = np.hstack([wx.flatten()])

    # Calculate DOS using histogram
    dos, bin_edges = np.histogram(frequencies, bins=num_bins, density=True)
    bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

    return bin_edges[1:], dos

def plot_dos_horizontal(axis, bin_centers, dos):
    #axis.barh(bin_centers, dos, height=bin_centers[1] - bin_centers[0], color="#1f77b4", alpha=1)
    axis.barh(bin_centers, dos, height=bin_centers[1] - bin_centers[0], color="black", edgecolor="white", linewidth=1, alpha=1)
    axis.set_xlabel('DOS')
    #axis.set_ylabel('$\omega$')
    axis.xaxis.set_major_locator(ticker.MultipleLocator(5))
    axis.set_xticks([])


def draw_graph_og(G, pos, m, k, box_size, periodic=False, ax=None):
    if ax is None:
        ax = plt.gca()
    if periodic:
        shifts = [(i * box_size, j * box_size) for i in (-1, 0, 1) for j in (-1, 0, 1)]
        for shift in shifts:
            for u, v in G.edges():
                linewidth = float((k[u] + k[v]) / 2) if hasattr(k, "__getitem__") else float(k)
                x1, y1 = np.array(pos[u]) + np.array(shift)
                x2, y2 = np.array(pos[v]) + np.array(shift)
                dx = x2 - x1
                dy = y2 - y1
                if dx > box_size / 2:
                    x2 -= box_size
                elif dx < -box_size / 2:
                    x2 += box_size
                if dy > box_size / 2:
                    y2 -= box_size
                elif dy < -box_size / 2:
                    y2 += box_size
                ax.plot([x1, x2], [y1, y2], 'k-', alpha=0.6, linewidth=linewidth)
        for shift in shifts:
            shifted_pos = {node: (pos[node][0] + shift[0], pos[node][1] + shift[1]) for node in G.nodes()}
            nx.draw_networkx_nodes(G, shifted_pos, node_size=5 * m, node_color="black", ax=ax)
    else:
        for u, v in G.edges():
            linewidth = float((k[u] + k[v]) / 2) if hasattr(k, "__getitem__") else float(k)
            x1, y1 = map(float, pos[u])
            x2, y2 = map(float, pos[v])
            ax.plot([x1, x2], [y1, y2], 'k-', alpha=0.6, linewidth=linewidth)
        nx.draw_networkx_nodes(G, pos, node_size=5 * m, node_color="black", ax=ax)
    ax.set_xlim(0.5, box_size + 0.5)
    ax.set_ylim(0.5, box_size + 0.5)
    ax.set_xticks([])
    ax.set_yticks([])

def draw_graph(G, pos, m, k, box_size, periodic=False, ax=None):
    box_size += 0.0
    if ax is None:
        ax = plt.gca()
    if periodic:
        shifts = [(i * (box_size), j * (box_size))for i in (-1,0,1) for j in (-1, 0, 1)]
        for shift in shifts:
            for u, v in G.edges():
                linewidth = float((k[u] + k[v]) / 2) if hasattr(k, "__getitem__") else float(k)
                x1, y1 = np.array(pos[u]) + np.array(shift)
                x2, y2 = np.array(pos[v]) + np.array(shift)
                dx = x2 - x1
                dy = y2 - y1
                if dx > box_size / 2:
                    x2 -= box_size
                elif dx < -box_size / 2:
                    x2 += box_size
                if dy > box_size / 2:
                    y2 -= box_size
                elif dy < -box_size / 2:
                    y2 += box_size
                ax.plot([x1, x2], [y1, y2], 'k-', alpha=0.6, linewidth=linewidth)
        for shift in shifts:
            shifted_pos = {node: (pos[node][0] + shift[0], pos[node][1] + shift[1]) for node in G.nodes()}
            nx.draw_networkx_nodes(G, shifted_pos, node_size=5 * m, node_color="black", ax=ax)
    else:
        for u, v in G.edges():
            linewidth = float((k[u] + k[v]) / 2) if hasattr(k, "__getitem__") else float(k)
            x1, y1 = map(float, pos[u])
            x2, y2 = map(float, pos[v])
            ax.plot([x1, x2], [y1, y2], 'k-', alpha=0.6, linewidth=linewidth)
        nx.draw_networkx_nodes(G, pos, node_size=5 * m, node_color="black", ax=ax)
    ax.set_xlim(1-0.5, box_size+0.5)
    ax.set_ylim(1-0.5, box_size+0.5)
    ax.set_xticks([])
    ax.set_yticks([])
```
